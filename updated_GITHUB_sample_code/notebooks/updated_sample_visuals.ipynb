# updated_sample_visuals

Executive-level visuals for roadmap, pipeline, and hypothesis approach.


```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
ROOT=Path('..').resolve()
OUT=ROOT/'reports'/'figures'/'executive_visuals'
OUT.mkdir(parents=True,exist_ok=True)
```

```python
phases=['Data Intake','Feature Engineering','Tree/Ensemble Modeling','Statistical Validation','Packaging','Review']
starts=np.array([0,1.2,3.0,4.8,6.3,7.3]); dur=np.array([1.2,1.8,1.8,1.5,1.0,0.9]); y=np.arange(len(phases))[::-1]
fig,ax=plt.subplots(figsize=(13,5.5))
colors=['#6C7A89','#415A77','#5E7D6A','#8A7A66','#6C7A89','#5E7D6A']
for yi,s,d,c,p in zip(y,starts,dur,colors,phases):
 ax.barh(yi,d,left=s,color=c,height=0.55); ax.text(s+d/2,yi,p,ha='center',va='center',color='white',fontsize=10)
ax.set_yticks([]); ax.set_xlim(0,8.5); ax.grid(axis='x',alpha=0.3); ax.set_title('Roadmap High Level')
fig.tight_layout(); fig.savefig(OUT/'roadmap_high_level.png',dpi=220); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(14,7)); ax.axis('off');
def box(x,y,w,h,t,s=''):
 r=FancyBboxPatch((x,y),w,h,boxstyle='round,pad=0.02,rounding_size=0.02',edgecolor='#CBD5E1',facecolor='#E9EEF3'); ax.add_patch(r); ax.text(x+w/2,y+h*0.62,t,ha='center',va='center',fontweight='bold');
 if s: ax.text(x+w/2,y+h*0.30,s,ha='center',va='center',fontsize=9,color='#6B7280')

box(0.03,0.62,0.17,0.22,'Input','sample csv'); box(0.24,0.62,0.17,0.22,'Features','time + lag'); box(0.45,0.62,0.17,0.22,'Models','boosting/bagging/stacking'); box(0.66,0.62,0.17,0.22,'Stats','p-values + CI'); box(0.85,0.62,0.12,0.22,'Outputs','reports')
for x1,x2 in [(0.20,0.24),(0.41,0.45),(0.62,0.66),(0.83,0.85)]:
 ax.add_patch(FancyArrowPatch((x1,0.73),(x2,0.73),arrowstyle='-|>',mutation_scale=12,color='#6B7280'))
ax.text(0.03,0.92,'Pipeline Architecture - High Level',fontsize=18,fontweight='bold')
fig.savefig(OUT/'pipeline_architecture_high_level.png',dpi=220,bbox_inches='tight'); plt.show()
```
